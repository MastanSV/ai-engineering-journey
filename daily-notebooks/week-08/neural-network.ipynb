```python
import numpy as np
import matplotlib.pyplot as plt
```

## Sigmoid

```python
x = np.linspace(-10, 10, 5)
print(x)
z = []
for item in x:
    res = 1 / (1 + np.exp(-item))
    z.append(res)

print(z)
```

```python
plt.plot(x, z)
plt.title('sigmoid function')
plt.xlabel('input : x')
plt.ylabel('output value : z')
```

As we can see in sigmoid function diagram, for the input value x, the output value is ranging in-between 0 & 1. and gradiant is disappearing at 0 & at 1. Major problems in this sigmoid function is, whatever the input value you give, there is no negative z value, for which it makes difficult for whether taking the sample into makes sense or not.

## tanh

```python
x = np.linspace(-10, 10, 5)


z = []

for item in x:
    res = np.tanh(item)
    z.append(res)

print(x)
print(z)
```

```python
plt.plot(x, z)
plt.title('tanh')
plt.xlabel('input value : x')
plt.ylabel('output value : z')
```

tanh solves the one of the problem which sigmoid has i.e., having negative output z values for the input x values, so we will get idea if the output is having negative value then we can consider it has less chances taken into the consideration. but the tanh still having the problem of gradiant is getting disappearing at -1 and 1. one of the good thing which tanh is that - it is zero centered.

## ReLU (Rectified Linear Unit)

In this ReLU, what is does is for the input value (x), if the x > 0, then it will be x, if x < 0, then output will be simply 0.

```python
x = np.linspace(-10, 10, 5)

z = []

for item in x:
    if item > 0:
        z.append(item)
    else:
        z.append(0)


print(x)
print(z)
```

```python
plt.plot(x, z)
plt.title('Rectified Linear Unit')
plt.xlabel('x')
plt.ylabel('z')
```

Rectified Linear Unit - which solves the problem of dying gradiant at the end values(In sigmoid 0&1, in tanh -1&1). But the ReLU has its own problem of dying ReLU which is nothing but when the input values simply contains all negative values the output will be just along the 0 which will not have any concrete conclusion.
